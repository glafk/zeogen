### Separate cifs according to zeolite code

```python
import os
import shutil


def clean_cif(input_file, output_file):
    # Open the input CIF file
    with open(input_file, 'r') as f:
        cif_data = f.readlines()

    # Filter out rows containing oxygen atoms (O)
    cleaned_cif_data = [line for line in cif_data if not line.startswith('O')]

    # Save the cleaned CIF data to the output file
    with open(output_file, 'w') as f:
        f.writelines(cleaned_cif_data)

def clean_all_in_directory(directory):
    for root, dirs, files in os.walk(directory):
        for filename in files:
            if filename.endswith(".cif"):
                input_cif_file = os.path.join(root, filename)
                output_cif_file = os.path.splitext(input_cif_file)[0] + '_clean.cif'
                clean_cif(input_cif_file, output_cif_file)
                os.remove(input_cif_file)  # Automatically delete the old file
                shutil.copy(output_cif_file, input_cif_file)  # Rename cleaned file to original name
                print(f"File '{input_cif_file}' has been cleaned.")

# Define the range of folder numbers
folder_range = range(1, 179)  # 1 through 178

# Root directory containing the numbered folders
root_dir = "C:/TUE/Thesis/zeogen/data/all_codes_dataset"  # Replace with the actual path
output_dir = root_dir

# List of reserved names in Windows
reserved_names = {"CON", "PRN", "AUX", "NUL", "COM1", "COM2", "COM3", "COM4", "COM5", "COM6", "COM7", "COM8", "COM9",
                  "LPT1", "LPT2", "LPT3", "LPT4", "LPT5", "LPT6", "LPT7", "LPT8", "LPT9"}

# Iterate over each folder
for folder_num in folder_range:
    folder_path = os.path.join(root_dir, str(folder_num))

    # Skip if the folder doesn't exist
    if not os.path.isdir(folder_path):
        print(f"Folder {folder_path} does not exist. Skipping...")
        continue

    # Iterate over each file in the folder
    for file_name in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file_name)

        # Skip if not a file
        if not os.path.isfile(file_path):
            continue

        # Extract the three-letter code from the file name
        try:
            code = file_name.split('_')[0]
        except IndexError:
            print(f"File name {file_name} in folder {folder_num} is not in the expected format. Skipping...")
            continue

        # Check for reserved names and modify if necessary
        if code.upper() in reserved_names:
            code = f"{code}_dir"

        # Create a directory for the code if it doesn't exist
        code_dir = os.path.join(output_dir, code)
        os.makedirs(code_dir, exist_ok=True)

        # Move the file to the new directory
        shutil.move(file_path, os.path.join(code_dir, file_name))

# Clean all CIF files in the newly created directories
for code_dir in os.listdir(output_dir):
    code_dir_path = os.path.join(output_dir, code_dir)
    if os.path.isdir(code_dir_path):
        clean_all_in_directory(code_dir_path)

print("File organization and cleaning complete.")
```
